```python
#Data visualisation#
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
plt.figure(figsize=(10,5))

sns.histplot(data= data , x='price', bins=30, color= 'green')
plt.title('Distribution of car price')
plt.xlabel(' Prices ')
plt.ylabel('Number of cars')
```

```python
sns.countplot(data=data , x='body-style')
plt.xticks(rotation=45)
plt.show
```

```python
sns.barplot(data=data, x='fuel-type', y='price')
plt.title('Average Price by Fuel Type')
plt.show()
```

```python
numeric_data = data.select_dtypes(include='number')
fig,ax=plt.subplots(figsize=(16,12))
sns.heatmap(numeric_data.corr(), annot=True, cmap='coolwarm',fmt='.3f', linewidths=0.5 , ax=ax)
plt.show()
```

```python
sns.scatterplot(data=data, x='horsepower', y='price')
plt.title('Price vs Horsepower')
plt.show()
```

```python
data.groupby('engine-type').price.mean().plot.bar()
```
